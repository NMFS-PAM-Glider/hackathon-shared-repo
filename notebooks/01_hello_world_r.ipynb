## 2. Which R is this?

Useful when something does not work and you need to say exactly what you were
running.

```r
cat("R version:", R.version.string, "\n")
cat("Running from:", R.home(), "\n")
```

## 3. Are the packages we need installed?

These packages are listed in `binder/install.R`, so Binder installs them when it
builds the environment. If any line says MISSING, the environment did not build
correctly.

```r
for (package in c("jsonlite", "readr", "dplyr", "ggplot2")) {
  status <- if (requireNamespace(package, quietly = TRUE)) "ok" else "MISSING"
  cat(sprintf("%-10s %s\n", package, status))
}
```
