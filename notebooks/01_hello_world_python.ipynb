## 3. Are the packages we need installed?

These four packages are listed in `binder/requirements.txt`, so Binder installs
them when it builds the environment. If any line says MISSING, the environment
did not build correctly.

```python
import importlib.util

for package in ["pandas", "requests", "matplotlib", "h5py"]:
    found = importlib.util.find_spec(package) is not None
    print(f"{package:12s} {'ok' if found else 'MISSING'}")
```
