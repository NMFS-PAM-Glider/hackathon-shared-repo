```python
with wave.open(str(wav_path), "rb") as wav:
    n_channels = wav.getnchannels()
    sample_width = wav.getsampwidth()
    sample_rate = wav.getframerate()
    n_frames = wav.getnframes()

    audio = np.frombuffer(
        wav.readframes(n_frames),
        dtype=np.int16
    )
```
